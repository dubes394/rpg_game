```python
class Bank:
    
    def __init__(self, balance):
        self.balance = balance
        
    def withdraw(self, amount):
        if amount < 0:
            raise Exception("amount cannot be negative")
        if self.balance < amount:
            raise Exception("not enough balance")
        self.balance = self.balance - amount
obj = Bank(10000)

try:
    obj.withdraw(14-999)
    
except Exception as e:
    print(e)
    
else:
    print(obj.balance)

```

```python
class MyException(Exception):
    
    def __init__(self, message):
        print(message)

class Bank:
    
    def __init__(self, balance):
        self.balance = balance
        
    def withdraw(self, amount):
        if amount < 0:
            raise MyException("amount cannot be negative")
        if self.balance < amount:
            raise MyException("not enough balance")
        self.balance = self.balance - amount
obj = Bank(10000)

try:
    obj.withdraw(14999)
    
except Exception as e:
    pass
    
else:
    print(obj.balance)
```

```python
class SecurityError(Exception):
    
    def __init__(self, message):
        print(message)
        
    def logout(self):
        print("Logout from all devices. Contact support")

class Google:
    
    def __init__(self, name, email, password, device):
        self.name = name
        self.device = device
        self.email = email
        self.password = password
        
    def login(self, email, password,device):
        
        if device != self.device:
            raise SecurityError("Device hacked")
        
        if email == self.email and password == self.password:
            print("welcome")
        else:
            print("login error")       
        
        
obj = Google('nitish', 'nitish@gmail.com', '123', 'ios')

try:
    obj.login('nitish@gmail.com', '123', 'linux')
except SecurityError as e:
    e.logout()
else:
    print(obj.name)
finally:
    print("DB is disconnected")    


```
