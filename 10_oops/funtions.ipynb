```python
def is_even(n):
    """_summary_
    this function returns if a given number is odd or even
    input = any valid integer
    output = odd/even
    created on = 16th nov 2022
    """
    
    if n % 2 == 0:
        return "even"
    else:
        return "odd"
```

```python
num = is_even(2)
print(num)
```

```python
for i in range(1,33):
    x= is_even(i)
    print(x)
```
