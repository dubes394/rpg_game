```python
class Customer:
    
    def __init__(self,name, gender, address):
        self.name = name
        self.gender = gender
        self.address = address
    
    def print_address(self):
        print(self.address.get_city(), self.gender, self.name)
        
class Address:
        
        def __init__(self, city, pin, state):
            self.__city = city
            self.pin = pin
            self.state = state
        
        def get_city(self):
            return self.__city
            
add1 = Address('bhp', 123, 'mp')
cust = Customer('nitish', 'male', add1)

```

```python
cust.print_address()
```
